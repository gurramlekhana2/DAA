```python

import pandas as pd

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}

df = pd.DataFrame(data)

# Compute covariance matrix
print("Covariance Matrix:")
print(df.cov())

# Compute correlation matrix
print("\nCorrelation Matrix:")
print(df.corr())

```

Covariance Matrix and Correlation Matrix

program uses Pandas to calculate the covariance matrix and correlation matrix for multiple attributes. First, it imports Pandas and creates a dataset containing attributes such as A, B, and C. The dataset is converted into a DataFrame using pd.DataFrame(). The df.cov() function calculates the covariance matrix, showing how each pair of attributes varies together, while the df.corr() function calculates the correlation matrix, indicating the strength and direction of the linear relationship between each pair of attributes. Finally, both matrices are displayed using the print() function.
