```python

import pandas as pd
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Marks': [20, 40, 50, 70, 90]
}
df = pd.DataFrame(data)
covariance = df['Study_Hours'].cov(df['Marks'])
correlation = df['Study_Hours'].corr(df['Marks'])

#AP24110010760
print("Covariance:", covariance)
print("Correlation:", correlation)

```

Compute Covariance and Correlation Between a Pair of Attributes



program uses the Pandas library to calculate covariance and correlation between two attributes. First, it imports Pandas and creates a dataset containing two attributes, such as study hours and marks. The dataset is converted into a DataFrame using pd.DataFrame(). The .cov() function calculates covariance to determine how the two attributes vary together, while the .corr() function calculates the Pearson correlation coefficient to measure the strength and direction of their linear relationship. Finally, the print() function displays the covariance and correlation values.
